# Factory Sponsor Three-Concept Scan Model

**Decision:** preserve all three direct sponsor YES votes while comparing only the bounded pre-software paid tests.

This notebook uses explicit assumptions, not observed sales. Official policy and product sources are listed in the companion report.

## Inputs and assumptions

Hours and BHD values are proposed test ranges. They are not market validation. One-time setup and per-order fulfillment are separated.

```python
concepts = [
    {"name": "Makeover Simulator", "setup_h": (12, 25), "price_bhd": (8, 25), "cash_bhd": (0, 30)},
    {"name": "TikTok Growth Operator", "setup_h": (15, 30), "price_bhd": (25, 75), "cash_bhd": (0, 40)},
    {"name": "YouTube Studio", "setup_h": (20, 40), "price_bhd": (40, 150), "cash_bhd": (0, 75)},
]
for item in concepts:
    midpoint_two_sales = sum(item["price_bhd"])
    print(f'{item["name"]}: setup {item["setup_h"][0]}-{item["setup_h"][1]}h, price BHD {item["price_bhd"][0]}-{item["price_bhd"][1]}, cash BHD {item["cash_bhd"][0]}-{item["cash_bhd"][1]}, midpoint revenue from 2 sales BHD {midpoint_two_sales:.1f}')
```

## Interpretation

A payment-count gate is insufficient. At Makeover's low price and high cash bound, two payments can still lose money before labor. Every live test therefore needs a settled-revenue and labor-floor gate. Media products should test two independent sellable wedges before extracting a shared kernel.

## Limitations

No demand, conversion, fulfillment time, platform approval, affiliate availability, or revenue was observed. The model exists only to make the first paid test falsifiable and prevent unsupported forecasts.
